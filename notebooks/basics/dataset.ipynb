```python
import os
import tarfile
import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt

def read_nifti(file_path):
    img = nib.load(file_path)
    return img.get_fdata()

id = 'IXI382-IOP-1135'
PD = read_nifti('d:/Downloads/IXI-PD/' + id + '-PD.nii.gz')
T1 = read_nifti('d:/Downloads/IXI-T1/' + id + '-T1.nii.gz')
T2 = read_nifti('d:/Downloads/IXI-T2/' + id + '-T2.nii.gz')

print('PD', PD.shape, PD.dtype)
print('T1', T1.shape, PD.dtype)
print('T2', T2.shape, PD.dtype)

index = 80
plt.imshow(PD[:,:,index])
plt.colorbar()
plt.show()

plt.imshow(T1[:,:,index])
plt.colorbar()
plt.show()

plt.imshow(T2[:,:,index])
plt.colorbar()
plt.show()
```
